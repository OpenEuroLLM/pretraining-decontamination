```python
import os
import pickle
from collections import defaultdict

def sum_matched_ngrams(pickle_dir):
    totals = defaultdict(int)

    for filename in os.listdir(pickle_dir):
        if not filename.endswith(".pkl") or "all_matched_ngrams" in filename:
            continue

        path = os.path.join(pickle_dir, filename)
        #print(path)
        with open(path, "rb") as f:
            data = pickle.load(f)

        # Support both dicts and objects
        if isinstance(data, dict):
            matched_ngrams = data.get("matched-ngrams", {})
        else:
            matched_ngrams = getattr(data, "matched-ngrams", {})

        for key, count in matched_ngrams.items():
            totals[key] += count

    return dict(totals)



pickle_path = "/scratch/project_462000963/nunezalc1/decontamination/nemotron/eng_Latn/matched_ngrams/"
result = sum_matched_ngrams(pickle_path)
print(len(result))
```

```python

def load_matched_ngram_keys(pickle_path):
    with open(pickle_path, "rb") as f:
        data = pickle.load(f)

    if isinstance(data, dict):
        matched_ngrams = data.get("matched-ngrams", {})
    else:
        matched_ngrams = getattr(data, "matched-ngrams", {})

    return set(matched_ngrams.keys())


def count_shared_keys(pickle_a, pickle_b):
    keys_a = load_matched_ngram_keys(pickle_a)
    keys_b = load_matched_ngram_keys(pickle_b)

    shared_keys = keys_a & keys_b
    return len(shared_keys), shared_keys


if __name__ == "__main__":
    pkl1 = "/scratch/project_462000963/nunezalc1/decontamination/nemotron/eng_Latn/matched_ngrams/matched_ngrams_1.0_high_actual_20.pkl"
    pkl2 = "/scratch/project_462000963/nunezalc1/decontamination/nemotron/eng_Latn/matched_ngrams/matched_ngrams_1.0_high_actual_22.pkl"

    count, shared = count_shared_keys(pkl1, pkl2)

    print(f"Shared keys: {count}")
```

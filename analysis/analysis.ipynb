```python
from collections import defaultdict
total = {}
def merge_level1(total, data):
    print(data)
    for field, value in data.items():
        if isinstance(value, defaultdict):
            if field not in total:
                total[field] = defaultdict(int)
            for k, v in value.items():
                total[field][k] += v
    return total



counts_a = defaultdict(lambda: defaultdict(int))
counts_a["matched_ngrams"]["apples"] = 3
counts_a["matched_ngrams"]["bananas"] = 5
counts_a["matched_ngrams"]["oranges"] = 2

# Second defaultdict
counts_b = defaultdict(lambda: defaultdict(int))
counts_b["matched_ngrams"]["red"] = 10
counts_b["matched_ngrams"]["blue"] = 4
counts_b["matched_ngrams"]["green"] = 7
counts_b["matched_ngrams"]["apples"] = 7

total = merge_level1(total, counts_a)
total = merge_level1(total, counts_b)

print(total)
```
